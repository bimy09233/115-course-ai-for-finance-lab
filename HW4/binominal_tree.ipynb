```python
import math
import numpy as np
```

```python
S0 = 50  # Current stock price
K = 40   # Strike price
r = 0.08  # Risk-free interest rate (annualized)
T = 2    # Time to expiration (in years)
v = 0.2  # Volatility of the stock (annualized)
```

```python
def binomial_tree(S0, K, r, T, v, steps=5):
    dt = T / steps
    u = math.exp(v * math.sqrt(dt))  # Up factor
    d = 1 / u                        # Down factor
    p = (math.exp(r * dt) - d) / (u - d)  # Risk-neutral probability        
    
    price = np.zeros((steps + 1, steps + 1))
    price[0, 0] = S0
    prob = np.zeros((steps + 1, steps + 1))
    prob[0, 0] = 1
    
    for col in range(1, steps + 1):
        for row in range(1, col + 1):
            price[row-1, col] = (price[row-1, col-1] * u)
            price[row, col] = (price[row-1, col-1] * d)
            
        # 從前一期每個節點，分別累加上漲與下跌路徑的機率
        for row in range(col):
            prob[row, col] += prob[row, col - 1] * p
            prob[row + 1, col] += prob[row, col - 1] * (1 - p)
            
    
    print(price)
    print(prob)
    
    expected_price = 0
    for row in range(steps + 1):
        if price[row, steps] > K:
            expected_price += ((price[row, steps]-K) * prob[row, steps])
            
    expected_price = expected_price * math.exp(-r * T)
    
    return expected_price

binomial_tree(S0, K, r, T, v, steps=5)
```
